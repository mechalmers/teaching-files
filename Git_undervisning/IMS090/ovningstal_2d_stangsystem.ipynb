## 2.29

<img src="images/2d_bar_system/u229.png" style="width:50%;">


```python
#kod från formelblad
def Kel(c,s, EA, L):
    return EA / L * sp.Matrix([[c**2, c*s, -c**2, -c*s],
                                [c*s, s**2, -c*s, -s**2],
                                [-c**2, -c*s, c**2, c*s],
                                [-c*s, -s**2, c*s, s**2]])

def assem(Kel,Ksystem,element_dof):
    for i in range(len(element_dof)):
        for j in range(len(element_dof)):
            Ksystem[element_dof[i], element_dof[j]] += Kel[i, j]
    return Ksystem
```

```python
#Lösning baserat på frihetsgrader etc enligt ovan
EA, L, a1, a2, a3, a4, a5, a6, a7, a8, P1, P2, P3, P4, P5, P6, P7, P8, P = sp.symbols('EA, L, a1, a2, a3, a4, a5, a6, a7, a8, P1, P2, P3, P4, P5, P6, P7, P8, P')

#stångelement 1
alpha1=0. 
s=np.sin(alpha1); c=np.cos(alpha1); 
L1=L
Kel1=Kel(c,s,EA,L1)
element_dof=[0,1,2,3] # pga Pythonindexering
Kmatris1=sp.zeros(8,8)
Kmatris1=assem(Kel1,Kmatris1,element_dof)


#stångelement 2
alpha2=np.arctan2(1.0,1.0)
s=np.sin(alpha2); c=np.cos(alpha2)
L2=sp.sqrt(2)*L
Kel2=Kel(c,s,EA,L2)
element_dof=[2,3,6,7]
Kmatris2=sp.zeros(8,8)
Kmatris2=assem(Kel2,Kmatris2,element_dof)

#stångelement 3
alpha3=np.arctan2(1.0,-1.0)
s=np.sin(alpha3); c=np.cos(alpha3)
L3=sp.sqrt(2)*L
Kel3=Kel(c,s,EA,L3)
element_dof=[2,3,4,5] 
Kmatris3=sp.zeros(8,8)
Kmatris3=assem(Kel3,Kmatris3,element_dof)


#stångelement 4
alpha4=0.0
s=np.sin(alpha4); c=np.cos(alpha4)
L4=2*L
Kel4=Kel(c,s,EA,L4)
element_dof=[4,5,6,7] 
Kmatris4=sp.zeros(8,8)
Kmatris4=assem(Kel4,Kmatris4,element_dof)



#Total systemstyvhetsmatris
Ksystem=Kmatris1+Kmatris2+Kmatris3+Kmatris4
#print("Systemstyvhetsmatris K:",Ksystem)

#Givna storheter
a1,a2,a4,a5,a6,a8=0,0,0,0,0,0
P3=0
P7=P

#Definiera vektorer
avektor=sp.Matrix([a1,a2,a3,a4,a5,a6,a7,a8])
Pvektor=sp.Matrix([P1,P2,P3,P4,P5,P6,P7,P8])

#Lös de obekanta
Sol=sp.solve(Ksystem*avektor-Pvektor, [a3,a7,P1,P2,P4,P5,P6,P8])

print(Sol)
a_res=avektor.subs(Sol)
print(f'a_res={a_res}')
print(f'a3={a_res[2]}')
print(f'a7={a_res[6]}')

#print('fac:',(8*(1+np.sqrt(2))/(8+5*np.sqrt(2))))
#print('facit:',4/(8+5*np.sqrt(2)))
```

```python
#från formelblad
def Kel(c,s, EA, L):
    return EA / L * sp.Matrix([[c**2, c*s, -c**2, -c*s],
                                [c*s, s**2, -c*s, -s**2],
                                [-c**2, -c*s, c**2, c*s],
                                [-c*s, -s**2, c*s, s**2]])

def assem(Kel,Ksystem,element_dof):
    for i in range(len(element_dof)):
        for j in range(len(element_dof)):
            Ksystem[element_dof[i], element_dof[j]] += Kel[i, j]
    return Ksystem
```

```python
#lösningen med geoemtri och dof enligt ovan
EA, L, a1, a2, a3, a4, a5, a6, P1, P2, P3, P4, P5, P6, P = sp.symbols('EA, L, a1, a2, a3, a4, a5, a6, P1, P2, P3, P4, P5, P6, P')

#stångelement 1
alpha1=-30.*np.pi/180 #
s=np.sin(alpha1); c=np.cos(alpha1); 
Kel1=Kel(c,s,EA,4*L)
element_dof=[0,1,2,3]
Kmatris1=sp.zeros(6,6)
Kmatris1=assem(Kel1,Kmatris1,element_dof)


#stångelement 2
alpha2=(30.+90.)*np.pi/180 
s=np.sin(alpha2); c=np.cos(alpha2);
Kel2=Kel(c,s,EA,2*sp.sqrt(3)*L)
element_dof=[4,5,0,1]
Kmatris2=sp.zeros(6,6)
Kmatris2=assem(Kel2,Kmatris2,element_dof)

#stångelement 3
alpha3=30.*np.pi/180 
s=np.sin(alpha3); c=np.cos(alpha3);
Kel3=Kel(c,s,EA,np.sqrt(3)*L)
element_dof=[4,5,2] 
Kmatris3=sp.zeros(6,6)
Kmatris3=assem(Kel3,Kmatris3,element_dof)


Kmatris=Kmatris1+Kmatris2+Kmatris3

#Givna storheter
a1, a3, a4, a6 = 0, 0, 0, 0

P2=-P
P5=0 

#Definiera vektorer
avektor=sp.Matrix([a1,a2,a3,a4,a5,a6])
Pvektor=sp.Matrix([P1,P2,P3,P4,P5,P6])

#Lös de obekanta
Sol=sp.solve(Kmatris*avektor-Pvektor, [a2,a5,P1,P3,P4,P6])

a_res=avektor.subs(Sol)
print(f'a_res={a_res}')
```
