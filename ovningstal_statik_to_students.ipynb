```python
#lösning b)
VA,HA,VB,F,M,a=sp.symbols('VA,HA,VB,F,M,a')

krafter_uppat=VA+VB-F*np.sin(np.pi/4) #Jämvikt vertikalt
krafter_sidled=HA+F*np.cos(np.pi/4) #Jämvikt sidled
Moment_A=M-F*np.sin(np.pi/4)*2*a-VB*3*a #Momentjämvikt kring A
losning=sp.solve([krafter_uppat,krafter_sidled,Moment_A],[VA,HA,VB])

print("HA=",losning[HA])
print("VA=",sp.expand(losning[VA]))
print("VB=",sp.expand(losning[VB]))
```
